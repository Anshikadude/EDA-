```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

fear_greed = pd.read_csv("/content/drive/MyDrive/fear_greed_index.csv")
trades = pd.read_csv("/content/drive/MyDrive/historical_data.csv")
```

```python
fear_greed.columns = (
    fear_greed.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

trades.columns = (
    trades.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

print("\nCleaned Fear & Greed Columns:\n", fear_greed.columns.tolist())
print("\nCleaned Trader Columns:\n", trades.columns.tolist())
```

```python
time_col = None

for col in trades.columns:
    if "time" in col.lower() or "date" in col.lower() or "timestamp" in col.lower():
        time_col = col
        break

print("\nDetected time column:", time_col)

# Convert to datetime
# First try milliseconds
trades["time"] = pd.to_datetime(trades[time_col], unit="ms", errors="coerce")

# If most values still NaT, try seconds
if trades["time"].isna().sum() > (0.5 * len(trades)):
    trades["time"] = pd.to_datetime(trades[time_col], unit="s", errors="coerce")

# If still NaT, try normal parsing
if trades["time"].isna().sum() > (0.5 * len(trades)):
    trades["time"] = pd.to_datetime(trades[time_col], errors="coerce")

print("\nMissing time values:", trades["time"].isna().sum())
```

```python
trades["date"] = trades["time"].dt.date
trades["hour"] = trades["time"].dt.hour
trades["minute"] = trades["time"].dt.minute

trades["date"] = pd.to_datetime(trades["date"])

print("\nAfter date-time split:")
print(trades[["time", "date", "hour", "minute"]].head())


# Drop rows without date (very important)
trades = trades.dropna(subset=["date"])

print("\nTrader data shape after cleaning:", trades.shape)
```

```python
fear_greed["date"] = pd.to_datetime(fear_greed["date"])
print("\nFear & Greed sample:")
print(fear_greed.head())
```

```python
# Candidate numeric columns
numeric_candidates = ['closedpnl', 'size', 'leverage', 'execution_price', 'start_position']

# Only keep columns that actually exist in your dataset
existing_numeric_cols = [col for col in numeric_candidates if col in trades.columns]

print("Existing numeric columns:", existing_numeric_cols)

# Convert to numeric
for col in existing_numeric_cols:
    trades[col] = pd.to_numeric(trades[col], errors="coerce")

# Fill NaN only for existing columns
trades[existing_numeric_cols] = trades[existing_numeric_cols].fillna(0)
```

```python
if "side" in trades.columns:
    trades["side"] = trades["side"].astype(str).str.upper()

    side_map = {
        "BUY": 1,
        "LONG": 1,
        "SELL": -1,
        "SHORT": -1
    }

    trades["side_num"] = trades["side"].map(side_map).fillna(0)
else:
    trades["side_num"] = 0
```

```python

if "size" in trades.columns:
    trades["abs_size"] = trades["size"].abs()
else:
    trades["abs_size"] = 0
```

```python
agg_dict = {}

if "closedpnl" in trades.columns:
    agg_dict["total_pnl"] = ("closedpnl", "sum")
    agg_dict["avg_pnl_per_trade"] = ("closedpnl", "mean")

if "abs_size" in trades.columns:
    agg_dict["total_volume"] = ("abs_size", "sum")

if "leverage" in trades.columns:
    agg_dict["avg_leverage"] = ("leverage", "mean")

# Count trades per day
if "account" in trades.columns:
    agg_dict["num_trades"] = ("account", "count")
else:
    agg_dict["num_trades"] = ("time", "count")

agg_dict["net_direction"] = ("side_num", "sum")


daily_metrics = trades.groupby("date").agg(**agg_dict).reset_index()

print("\nDaily Metrics Preview:")
print(daily_metrics.head())


# Save daily trader metrics
daily_metrics.to_csv("daily_trader_metrics.csv", index=False)
```

```python

merged = pd.merge(
    daily_metrics,
    fear_greed[["date", "classification"]],
    on="date",
    how="inner"
)

print("\nMerged dataset:")
print(merged.head())

merged.to_csv("merged_sentiment_trades.csv", index=False)
```

```python
available_cols = [c for c in ['total_pnl','total_volume','avg_leverage','num_trades','net_direction'] if c in merged.columns]

print("Usable columns:", available_cols)

print(
    merged
    .groupby('classification')[available_cols]
    .agg(['mean','median'])
)

rename_map = {
    'sum_pnl': 'total_pnl',
    'pnl': 'total_pnl',
    'volume': 'total_volume',
    'vol': 'total_volume',
    'mean_leverage': 'avg_leverage',
    'leverage_mean': 'avg_leverage'
}

merged = merged.rename(columns=rename_map)
print(merged.columns.tolist())


```
